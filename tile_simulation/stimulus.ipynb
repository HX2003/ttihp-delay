```python
def make_pwl(bits, v_high='1.2', v_low='0', period_ns=20, rise_ns=0.1):
    # Convert tuple/list to a single string
    if isinstance(bits, (tuple, list)):
        bits = ''.join(bits)

    pts = []
    for i, b in enumerate(bits):
        v = v_high if b == '1' else v_low
        t = i * period_ns
        if i == 0:
            pts.append((t, v))
        else:
            prev_v = v_high if bits[i-1] == '1' else v_low
            if v != prev_v:
                pts.append((t, prev_v))
                pts.append((t + rise_ns, v))
    pts.append((len(bits) * period_ns, v_high if bits[-1] == '1' else v_low))
    return "PWL(" + " ".join(f"{t:.1f}n {v}" for t, v in pts) + ")"

# It first performs initialization procedure,
# then writes the following
# write_reg(0, 2)
# write_reg(1, 6)
# write_reg(2, 7)
# write_reg(3, 18)
# write_reg(4, 31)
print("STIMULUS_rst_n     " + make_pwl(('000', '111', '111', '111', '111', '111',)))
print("STIMULUS_rst_clk   " + make_pwl(('010', '010', '010', '010', '010', '010')))
print("STIMULUS_reg_addr0 " + make_pwl(('000', '000', '111', '000', '111', '000')))
print("STIMULUS_reg_addr1 " + make_pwl(('000', '000', '000', '111', '111', '000')))
print("STIMULUS_reg_addr2 " + make_pwl(('000', '000', '000', '000', '000', '111')))
print("STIMULUS_reg_val0  " + make_pwl(('000', '000', '000', '111', '000', '111')))
print("STIMULUS_reg_val1  " + make_pwl(('000', '111', '111', '111', '111', '111')))
print("STIMULUS_reg_val2  " + make_pwl(('000', '000', '111', '111', '000', '111')))
print("STIMULUS_reg_val3  " + make_pwl(('000', '000', '000', '000', '000', '111')))
print("STIMULUS_reg_val4  " + make_pwl(('000', '000', '000', '000', '111', '111')))
```
